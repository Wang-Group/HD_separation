# Pareto front evolution (batch-by-batch) with polygon enclosure

- Objectives are **maximized**:
  - `separation_coefficient` (higher is better)
  - `lsv_overpotential_-0.01` (treated as "voltage"; higher / less negative is better)

- Samples are grouped into batches of 4 by index (`0–3` = batch 1, `4–7` = batch 2, ...).
- Reference point for enclosure is the **dataset minima** of both objectives (used for geometry; not plotted).
- For each cumulative batch, we draw:
  - Pareto front line
  - A single **polygon enclosure** linked between the reference and the Pareto front
  - A boundary line wrapping the polygon

Styling:
- Rainbow colormap
- Area alpha = **0.05**
- Line alpha = **0.5**


```python
import json
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

from shapely.geometry import Polygon, box, MultiPolygon, GeometryCollection
from shapely.ops import unary_union

# =========================
# ---- Input / Settings ----
# =========================
for json_path in [Path("initial_sampling_Co_Mo.json"), Path("initial_sampling_Co_Ni.json"), Path("initial_sampling_Co_V.json"), Path("initial_sampling_Fe_Ni.json")]:
    voltage_key = "lsv_overpotential_-0.01"
    sep_key = "separation_coefficient"

    FIGSIZE = (8, 8)
    S_SCATTER = 100
    SCATTER_ALPHA = 0.50

    AREA_ALPHA = 0.20         # Δ新增区域透明度

    HV_LINE_ALPHA_EARLY = 0.80
    HV_LINE_ALPHA_LAST  = 0.80
    HV_LINE_LW = 2.0

    # 只在最后一波画 Pareto front 线
    PARETO_LINE_ALPHA_LAST = 0.80
    PARETO_LINE_LW = 2.5
    PARETO_MARKERSIZE = 10*2

    # 每一波 Pareto 点（只上色）
    PARETO_POINT_SIZE = 90*2
    PARETO_POINT_ALPHA = 0.85

    REVERSE_RAINBOW = True

    # =========================
    # ---- Load / DataFrame ----
    # =========================
    data = json.loads(json_path.read_text())

    def dictcol_to_series(d):
        s = pd.Series(d)
        s.index = s.index.astype(int)
        return s.sort_index()

    df = pd.DataFrame({
        "id": dictcol_to_series(data[voltage_key]).index,
        "voltage": dictcol_to_series(data[voltage_key]).values,
        "separation": dictcol_to_series(data[sep_key]).values,
    }).dropna(subset=["voltage", "separation"]).reset_index(drop=True)

    df["id"] = df["id"].astype(int)
    df = df.sort_values("id").reset_index(drop=True)

    df["batch"] = (df["id"] // 4) + 1
    max_batch = int(df["batch"].max())

    # Reference point = dataset minima (used for HV; not plotted)
    x_ref = float(df["voltage"].min())
    y_ref = float(df["separation"].min())

    # =========================
    # ---- Pareto front (maximize both) ----
    # =========================
    def pareto_front_max(df_in: pd.DataFrame, x="voltage", y="separation"):
        s = df_in.sort_values([x, y], ascending=[False, False]).reset_index(drop=True)
        best_y = -np.inf
        keep = []
        for _, r in s.iterrows():
            if r[y] > best_y + 1e-12:
                keep.append(True)
                best_y = r[y]
            else:
                keep.append(False)
        front = s.loc[keep].copy()
        return front.sort_values(x, ascending=True).reset_index(drop=True)

    # =========================
    # ---- Shapely helpers ----
    # =========================
    def fix_geom(g):
        if g is None:
            return Polygon()
        if not g.is_valid:
            g = g.buffer(0)
        return g

    def iter_polygons(g):
        if g is None or g.is_empty:
            return
        gt = g.geom_type
        if gt == "Polygon":
            yield g
        elif gt == "MultiPolygon":
            yield from g.geoms
        elif gt == "GeometryCollection":
            for gg in g.geoms:
                yield from iter_polygons(gg)

    def draw_geom_boundary(ax, geom, color, alpha=0.3, lw=2.0):
        for p in iter_polygons(geom):
            x, y = p.exterior.xy
            ax.plot(x, y, color=color, alpha=alpha, linewidth=lw)

    def fill_geom(ax, geom, color, alpha=0.05):
        for p in iter_polygons(geom):
            x, y = p.exterior.xy
            ax.fill(x, y, color=color, alpha=alpha, linewidth=0)

    # =========================
    # ---- Hypervolume region (2D) ----
    # =========================
    def hv_union_geom(front: pd.DataFrame, x_ref: float, y_ref: float,
                    x="voltage", y="separation"):
        rects = []
        for _, r in front.iterrows():
            xi = float(r[x]); yi = float(r[y])
            if xi <= x_ref or yi <= y_ref:
                continue
            rects.append(box(x_ref, y_ref, xi, yi))
        if not rects:
            return Polygon()
        return fix_geom(unary_union(rects))

    # =========================
    # ---- Build cumulative fronts + HV ----
    # =========================
    fronts = {}
    hv_geoms = {}
    for b in range(1, max_batch + 1):
        cum = df[df["batch"] <= b].copy()
        fronts[b] = pareto_front_max(cum)
        hv_geoms[b] = hv_union_geom(fronts[b], x_ref, y_ref)

    # =========================
    # ---- Plot ----
    # =========================
    fig, ax = plt.subplots(figsize=FIGSIZE)
    ax.scatter(df["voltage"], df["separation"], s=S_SCATTER, alpha=SCATTER_ALPHA, label="All Samples", zorder = 10)

    # Colors
    cmap = plt.cm.rainbow
    colors = [cmap(i / max(1, (max_batch - 1))) for i in range(max_batch)]
    if REVERSE_RAINBOW:
        colors = colors[::-1]

    # 1) Δ新增 hypervolume 填充（避免重叠糊）
    covered = fix_geom(Polygon())
    for b in range(1, max_batch + 1):
        c = colors[b-1]
        hv_b = hv_geoms[b]
        delta = fix_geom(hv_b.difference(covered))
        covered = fix_geom(unary_union([covered, hv_b]))
        fill_geom(ax, delta, c, alpha=AREA_ALPHA)

    # 2) 前几波：只画 HV 边界线（淡）
    for b in range(1, max_batch):
        c = colors[b-1]
        draw_geom_boundary(ax, hv_geoms[b], c, alpha=HV_LINE_ALPHA_EARLY, lw=HV_LINE_LW)

    # 3) 最后一波：画 HV 边界线（强调）
    last_color = colors[max_batch - 1]
    draw_geom_boundary(ax, hv_geoms[max_batch], last_color, alpha=HV_LINE_ALPHA_LAST, lw=HV_LINE_LW)

    # 4) 每一波：把 Pareto front 的点画出来（只给点颜色，不画线）
    for b in range(1, max_batch + 1):
        f = fronts[b]
        c = colors[b-1]
        ax.scatter(
            f["voltage"], f["separation"],
            s=PARETO_POINT_SIZE,
            color=c,
            edgecolors="black",
            linewidths=0.8,      # ✅ 这里
            alpha=PARETO_POINT_ALPHA,
            zorder=100 - b,
            label=f"Batch {b}"
        )
        #  # 虚线连接（dashed）
        # ax.plot(
        #     f["voltage"], f["separation"],
        #     linestyle="--",
        #     linewidth=1.5,
        #     color=c,
        #     alpha=0.6,
        #     zorder=99 - b
        # )

    # 5) 仅最后一波：画 Pareto front 线
    f_last = fronts[max_batch]
    ax.plot(
        f_last["voltage"], f_last["separation"],
        linewidth=PARETO_LINE_LW,
        markersize=PARETO_MARKERSIZE,
        color=last_color,
        alpha=PARETO_LINE_ALPHA_LAST,
        label=f"Pareto Front",
        linestyle="--",
    )

    # ---- axis formatting (沿用你之前风格) ----
    from matplotlib.ticker import MultipleLocator
    ax.yaxis.set_major_locator(MultipleLocator(1))
    ax.xaxis.set_major_locator(MultipleLocator(0.04))

    ax.tick_params(axis="both", which="both", labelsize=24)
    for t in ax.get_xticklabels() + ax.get_yticklabels():
        t.set_fontname("Arial")
        t.set_fontweight("bold")

    ax.set_ylabel("Separation Coefficient (@ -50 mA/cm²)", fontname="Arial", fontweight="bold", fontsize=28)
    ax.set_xlabel("Potential (@ -10 mA/cm², vs. RHE, V)", fontname="Arial", fontweight="bold", fontsize=28)

    plt.tight_layout()
    plt.show()

    out_png = Path(str(json_path).replace(".json", ".png"))
    fig.savefig(out_png, dpi=600)
    print("Saved:", out_png.resolve())
    print(f"Reference point used (not plotted): x_ref={x_ref:.6f}, y_ref={y_ref:.6f}")
    print(f"Final hypervolume area: {hv_geoms[max_batch].area:.6f}")

    import numpy as np
    import pandas as pd
    import matplotlib.pyplot as plt
    from matplotlib.patches import Rectangle

    features = ["lsv_overpotential_-0.01", "separation_coefficient"]
    data = pd.DataFrame(data).copy()

    # ---- build indices: 3, 7, 8, 12, 16, 20, ...
    idx = [3, 7, 8] + list(range(12, len(data), 4))
    idx = [i for i in idx if 0 <= i < len(data)]  # safe guard

    pts = data.iloc[idx][features].apply(pd.to_numeric, errors="coerce").dropna(how="any")

    x = pts[features[0]].to_numpy()
    y = pts[features[1]].to_numpy()

    mean_x, mean_y = np.mean(x), np.mean(y)
    std_x  = np.std(x, ddof=1) if len(x) > 1 else np.nan
    std_y  = np.std(y, ddof=1) if len(y) > 1 else np.nan

    plt.figure(figsize=(6.5, 6.0))
    plt.scatter(x, y, s=60, alpha=0.85, label=f"points (n={len(pts)})")

    plt.errorbar(
        mean_x, mean_y,
        xerr=std_x if np.isfinite(std_x) else None,
        yerr=std_y if np.isfinite(std_y) else None,
        fmt="o", markersize=10, capsize=5, linewidth=2,
        label="mean ± 1 std (x,y)"
    )

    print(f"Picked iloc rows: {idx}")
    print(f"Mean ± 1 std:\n  x: {mean_x:.6f} ± {std_x:.6f}\n  y: {mean_y:.6f} ± {std_y:.6f}")

    if np.isfinite(std_x) and np.isfinite(std_y):
        rect = Rectangle((mean_x - std_x, mean_y - std_y),
                        2*std_x, 2*std_y,
                        fill=False, linewidth=2, alpha=0.8)
        plt.gca().add_patch(rect)

    plt.xlabel(features[0])
    plt.ylabel(features[1])
    plt.legend()
    plt.tight_layout()
    plt.show()
```

```python
import json
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
from shapely.geometry import Polygon, box
from shapely.ops import unary_union

# -------------------------
# keys (same as before)
# -------------------------
voltage_key = "lsv_overpotential_-0.01"
sep_key     = "separation_coefficient"

# -------------------------
# style
# -------------------------
FIGSIZE = (8, 9)
LINE_LW = 2.5
LINE_ALPHA = 0.85
SCATTER_S = 300
SCATTER_ALPHA = 0.95
HIGHLIGHT_S = 300
REVERSE_RAINBOW = True

# choose what to plot:
#   "gain"       : HV(b) - HV(1)   (starts at 0)
#   "absolute"   : HV(b)
#   "normalized" : (HV(b)-HV(1)) / (HV(last)-HV(1))   (0→1)
HV_MODE = "gain"


# -------------------------
# helpers (same logic as before)
# -------------------------
def dictcol_to_series(d):
    s = pd.Series(d)
    s.index = s.index.astype(int)
    return s.sort_index()

import matplotlib.transforms as mtransforms

def shiny_ball(ax, x, y, s=140, color=None, edgecolor="black", lw=1.0,
               highlight_alpha=0.35, zorder=20):
    """
    Draw a 'shiny' ball by layering:
      - base filled circle
      - small white highlight slightly offset in screen points
    """
    if color is None:
        color = plt.rcParams["axes.prop_cycle"].by_key()["color"][0]

    # base ball
    ax.scatter([x], [y], s=s, color=color, edgecolors=edgecolor, linewidths=lw, zorder=zorder)

    # highlight (offset in points so it looks consistent at any scale)
    fig = ax.figure
    offset = mtransforms.ScaledTranslation(-2/72., 2/72., fig.dpi_scale_trans)  # (-2pt, +2pt)
    ax.scatter([x], [y],
               s=s * 0.22,
               color="white",
               alpha=highlight_alpha,
               linewidths=0,
               transform=ax.transData + offset,
               zorder=zorder + 10)

def pareto_front_max(df_in: pd.DataFrame, x="voltage", y="separation"):
    s = df_in.sort_values([x, y], ascending=[False, False]).reset_index(drop=True)
    best_y = -np.inf
    keep = []
    for _, r in s.iterrows():
        if r[y] > best_y + 1e-12:
            keep.append(True)
            best_y = r[y]
        else:
            keep.append(False)
    front = s.loc[keep].copy()
    return front.sort_values(x, ascending=True).reset_index(drop=True)

def fix_geom(g):
    if g is None:
        return Polygon()
    if not g.is_valid:
        g = g.buffer(0)
    return g

def hv_union_geom(front: pd.DataFrame, x_ref: float, y_ref: float,
                  x="voltage", y="separation"):
    rects = []
    for _, r in front.iterrows():
        xi = float(r[x]); yi = float(r[y])
        if xi <= x_ref or yi <= y_ref:
            continue
        rects.append(box(x_ref, y_ref, xi, yi))
    if not rects:
        return Polygon()
    return fix_geom(unary_union(rects))

def compute_hv_series(json_path: Path):
    data = json.loads(json_path.read_text())

    df = pd.DataFrame({
        "id": dictcol_to_series(data[voltage_key]).index,
        "voltage": dictcol_to_series(data[voltage_key]).values,
        "separation": dictcol_to_series(data[sep_key]).values,
    }).dropna(subset=["voltage", "separation"]).reset_index(drop=True)

    df["id"] = df["id"].astype(int)
    df = df.sort_values("id").reset_index(drop=True)
    df["batch"] = (df["id"] // 4) + 1
    max_batch = int(df["batch"].max())

    x_ref = float(df["voltage"].min())
    y_ref = float(df["separation"].min())

    hv = []
    batches = np.arange(1, max_batch + 1, dtype=int)
    for b in batches:
        cum = df[df["batch"] <= b].copy()
        front = pareto_front_max(cum)
        hv_geom = hv_union_geom(front, x_ref, y_ref)
        hv.append(float(fix_geom(hv_geom).area))

    hv = np.asarray(hv, dtype=float)

    if HV_MODE == "absolute":
        y = hv
    elif HV_MODE == "gain":
        y = hv - hv[0]
    elif HV_MODE == "normalized":
        denom = (hv[-1] - hv[0])
        y = (hv - hv[0]) / denom if abs(denom) > 1e-12 else np.zeros_like(hv)
    else:
        raise ValueError("HV_MODE must be one of: 'gain', 'absolute', 'normalized'")

    return batches, y, hv


# -------------------------
# plotting per panel (font/style aligned to your 2nd script)
# -------------------------
def plot_hv_panel(ax, json_path: Path, sys_name: str):
    batches, y, hv_abs = compute_hv_series(json_path)

    # one unified color for line + balls
    main_color = plt.rcParams["axes.prop_cycle"].by_key()["color"][0]
    main_color = "#FF42AD"   # RGB = (255, 66, 173)

    # one line style (no rainbow, no multi scatters)
    ax.plot(
        batches, y,
        linewidth=LINE_LW,
        alpha=LINE_ALPHA,
        color=main_color,
        zorder=10
    )

    # shiny balls for each batch
    for b in batches:
        shiny_ball(ax, b, y[b - 1], s=SCATTER_S*1.2, color=main_color, lw=1.0, zorder=20)

    # highlight batch 1 + last (bigger shiny balls)
    shiny_ball(ax, batches[0],  y[0],  s=HIGHLIGHT_S, color=main_color, lw=0.0, zorder=30)
    shiny_ball(ax, batches[-1], y[-1], s=HIGHLIGHT_S, color=main_color, lw=0.0, zorder=31)

    # system name style like your other figure (xlabel, not title)
    ax.set_xlabel(sys_name, fontname="Arial", fontweight="bold", fontsize=28)
    ax.set_ylabel("")

    # x ticks: integer batches
    ax.set_xticks(batches)

    # tick styling
    ax.tick_params(axis="both", which="both", labelsize=24)
    for t in ax.get_xticklabels() + ax.get_yticklabels():
        t.set_fontname("Arial")
        t.set_fontweight("bold")
        
    ax.margins(x=0.05, y=0.10)
    FRAME_LW = 2.5  # <- change this

    for side in ["left", "right", "top", "bottom"]:
        ax.spines[side].set_linewidth(FRAME_LW)


# -------------------------
# 1×4 layout + shared labels (system name as xlabel, not title)
# -------------------------
systems = [
    (Path("initial_sampling_Co_Ni.json"), "Co-Ni"),
    (Path("initial_sampling_Co_Mo.json"), "Co-Mo"),
    (Path("initial_sampling_Fe_Ni.json"), "Fe-Ni"),
    (Path("initial_sampling_Co_V.json"),  "Co-V"),
]

fig, axs = plt.subplots(
    1, len(systems),
    figsize=(FIGSIZE[0]*len(systems), FIGSIZE[1])
)

for ax, (p, sys_name) in zip(axs, systems):
    plot_hv_panel(ax, p, sys_name)
    ax.set_title(sys_name, fontname="Arial", fontweight="bold", fontsize=30, pad=14)
    ax.set_xlabel("")   # remove bottom label
    ax.set_ylabel("")

# layout like your 2nd script
fig.tight_layout(rect=[0.04, 0.02, 1.04, 1.04])

import matplotlib.transforms as mtransforms
bbox = mtransforms.Bbox.union([ax.get_position() for ax in axs])
y_center = 0.5 * (bbox.y0 + bbox.y1)
x_center = 0.5 * (bbox.x0 + bbox.x1)
x_left   = bbox.x0

y_label = {
    "gain": "Improvement of Hypervolume",
    "absolute": "Hypervolume",
    "normalized": "Normalized Hypervolume Gain (0–1)"
}[HV_MODE]

pad = 0.02
fig.text(
    x_left - 2*pad, y_center,
    y_label,
    rotation="vertical",
    va="center", ha="center",
    fontname="Arial", fontweight="bold", fontsize=30
)

fig.text(
    x_center, 0,   # match your 2nd script (y=0)
    "Batch Number",
    va="center", ha="center",
    fontname="Arial", fontweight="bold", fontsize=30
)

plt.show()
fig.savefig("four_systems_hv_increase_row.png", dpi=600, bbox_inches="tight", pad_inches=0.02)
```

```python
import json
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
from shapely.geometry import Polygon, box
from shapely.ops import unary_union
import matplotlib.transforms as mtransforms

# -------------------------
# keys
# -------------------------
voltage_key = "lsv_overpotential_-0.01"
sep_key     = "separation_coefficient"

# -------------------------
# systems (4 columns)
# -------------------------
systems = [
    (Path("initial_sampling_Co_Ni.json"), "Co-Ni"),
    (Path("initial_sampling_Co_Mo.json"), "Co-Mo"),
    (Path("initial_sampling_Fe_Ni.json"), "Fe-Ni"),
    (Path("initial_sampling_Co_V.json"),  "Co-V"),
]

# -------------------------
# TOP row style (Pareto + HV area)
# -------------------------
TOP_FIGSIZE_PER_PANEL = (8, 9)  # (w,h) per panel in inches
S_SCATTER = 100
SCATTER_ALPHA = 0.95
AREA_ALPHA = 0.05
HV_LINE_ALPHA_EARLY = 0.80
HV_LINE_ALPHA_LAST  = 0.80
HV_LINE_LW = 2.0
PARETO_LINE_ALPHA_LAST = 0.80
PARETO_LINE_LW = 2.5
PARETO_MARKERSIZE = 20
PARETO_POINT_SIZE = 180
PARETO_POINT_ALPHA = 0.85
REVERSE_RAINBOW = True
FRAME_LW = 2.5

# -------------------------
# BOTTOM row style (HV vs batch)
# -------------------------
LINE_LW = 2.5
LINE_ALPHA = 0.85
SCATTER_S = 300
SCATTER_ALPHA = 0.95
HIGHLIGHT_S = 300

# choose what to plot:
#   "gain"       : HV(b) - HV(1)   (starts at 0)
#   "absolute"   : HV(b)
#   "normalized" : (HV(b)-HV(1)) / (HV(last)-HV(1))   (0→1)
HV_MODE = "gain"

# -------------------------
# helpers
# -------------------------
def dictcol_to_series(d):
    s = pd.Series(d)
    s.index = s.index.astype(int)
    return s.sort_index()

def load_system_df(json_path: Path):
    data = json.loads(json_path.read_text())
    
    df = pd.DataFrame({
        "id": dictcol_to_series(data[voltage_key]).index,
        "voltage": dictcol_to_series(data[voltage_key]).values + 0.0592*14 + 0.098,
        "separation": dictcol_to_series(data[sep_key]).values,
    }).dropna(subset=["voltage", "separation"]).reset_index(drop=True)

    df["id"] = df["id"].astype(int)
    df = df.sort_values("id").reset_index(drop=True)
    df["batch"] = (df["id"] // 4) + 1
    max_batch = int(df["batch"].max())

    x_ref = float(df["voltage"].min())
    y_ref = float(df["separation"].min())
    return df, max_batch, x_ref, y_ref

def pareto_front_max(df_in: pd.DataFrame, x="voltage", y="separation"):
    s = df_in.sort_values([x, y], ascending=[False, False]).reset_index(drop=True)
    best_y = -np.inf
    keep = []
    for _, r in s.iterrows():
        if r[y] > best_y + 1e-12:
            keep.append(True)
            best_y = r[y]
        else:
            keep.append(False)
    front = s.loc[keep].copy()
    return front.sort_values(x, ascending=True).reset_index(drop=True)

def fix_geom(g):
    if g is None:
        return Polygon()
    if not g.is_valid:
        g = g.buffer(0)
    return g

def iter_polygons(g):
    if g is None or g.is_empty:
        return
    gt = g.geom_type
    if gt == "Polygon":
        yield g
    elif gt == "MultiPolygon":
        yield from g.geoms
    elif gt == "GeometryCollection":
        for gg in g.geoms:
            yield from iter_polygons(gg)

def draw_geom_boundary(ax, geom, color, alpha=0.3, lw=2.0):
    for p in iter_polygons(geom):
        x, y = p.exterior.xy
        ax.plot(x, y, color=color, alpha=alpha, linewidth=lw)

def fill_geom(ax, geom, color, alpha=0.05):
    for p in iter_polygons(geom):
        x, y = p.exterior.xy
        ax.fill(x, y, color=color, alpha=alpha, linewidth=0)

def hv_union_geom(front: pd.DataFrame, x_ref: float, y_ref: float,
                  x="voltage", y="separation"):
    rects = []
    for _, r in front.iterrows():
        xi = float(r[x]); yi = float(r[y])
        if xi <= x_ref or yi <= y_ref:
            continue
        rects.append(box(x_ref, y_ref, xi, yi))
    if not rects:
        return Polygon()
    return fix_geom(unary_union(rects))

def shiny_ball(ax, x, y, s=140, color=None, edgecolor="black", lw=1.0,
               highlight_alpha=0.35, zorder=20):
    if color is None:
        color = plt.rcParams["axes.prop_cycle"].by_key()["color"][0]

    ax.scatter([x], [y], s=s, color=color, edgecolors=edgecolor, linewidths=lw, zorder=zorder)

    fig = ax.figure
    offset = mtransforms.ScaledTranslation(-2/72., 2/72., fig.dpi_scale_trans)
    ax.scatter([x], [y],
               s=s * 0.22,
               color="white",
               alpha=highlight_alpha,
               linewidths=0,
               transform=ax.transData + offset,
               zorder=zorder + 10)

# -------------------------
# TOP row plotting
# -------------------------
def plot_system(ax, json_path: Path):
    df, max_batch, x_ref, y_ref = load_system_df(json_path)
    fronts, hv_geoms = {}, {}
    for b in range(1, max_batch + 1):
        cum = df[df["batch"] <= b].copy()
        fronts[b] = pareto_front_max(cum)
        hv_geoms[b] = hv_union_geom(fronts[b], x_ref, y_ref)

    cmap = plt.cm.rainbow
    colors = [cmap(i / max(1, (max_batch - 1))) for i in range(max_batch)]
    if REVERSE_RAINBOW:
        colors = colors[::-1]

    # scatter all points colored by batch
    for b in range(1, max_batch + 1):
        c = colors[b - 1]
        dfb = df[df["batch"] == b]
        ax.scatter(dfb["voltage"], dfb["separation"],
                   s=S_SCATTER, color=c, alpha=SCATTER_ALPHA, edgecolors="black", linewidths=0.8, zorder=110)

    c1 = colors[0]
    cl = colors[max_batch - 1]

    hv1 = fix_geom(hv_geoms[1])
    hvl = fix_geom(hv_geoms[max_batch])

    fill_geom(ax, hv1, c1, alpha=AREA_ALPHA)
    if max_batch > 1:
        delta_last = fix_geom(hvl.difference(hv1))
        fill_geom(ax, delta_last, cl, alpha=AREA_ALPHA)

    draw_geom_boundary(ax, hv1, c1, alpha=HV_LINE_ALPHA_EARLY, lw=HV_LINE_LW)
    if max_batch > 1:
        draw_geom_boundary(ax, hvl, cl, alpha=HV_LINE_ALPHA_LAST, lw=HV_LINE_LW)

    # Pareto front for batch 1 + last batch
    f1 = fronts[1]
    ax.scatter(f1["voltage"], f1["separation"],
               s=PARETO_POINT_SIZE, color=c1,
               edgecolors="black", linewidths=0.8,
               alpha=PARETO_POINT_ALPHA, zorder=120)
    ax.plot(f1["voltage"], f1["separation"],
            linewidth=PARETO_LINE_LW, markersize=PARETO_MARKERSIZE,
            color=c1, alpha=PARETO_LINE_ALPHA_LAST, linestyle="--")

    f_last = fronts[max_batch]
    ax.scatter(f_last["voltage"], f_last["separation"],
               s=PARETO_POINT_SIZE, color=cl,
               edgecolors="black", linewidths=0.8,
               alpha=PARETO_POINT_ALPHA, zorder=130)
    ax.plot(f_last["voltage"], f_last["separation"],
            linewidth=PARETO_LINE_LW, markersize=PARETO_MARKERSIZE,
            color=cl, alpha=PARETO_LINE_ALPHA_LAST, linestyle="--")

    # ticks + frame
    from matplotlib.ticker import MultipleLocator
    ax.yaxis.set_major_locator(MultipleLocator(1))
    ax.xaxis.set_major_locator(MultipleLocator(0.05))
    ax.tick_params(axis="both", which="both", labelsize=24)
    for t in ax.get_xticklabels() + ax.get_yticklabels():
        t.set_fontname("Arial")
        t.set_fontweight("bold")
    for side in ["left", "right", "top", "bottom"]:
        ax.spines[side].set_linewidth(FRAME_LW)

# -------------------------
# BOTTOM row plotting
# -------------------------
def compute_hv_series(json_path: Path):
    df, max_batch, x_ref, y_ref = load_system_df(json_path)

    hv = []
    batches = np.arange(1, max_batch + 1, dtype=int)
    for b in batches:
        cum = df[df["batch"] <= b].copy()
        front = pareto_front_max(cum)
        hv_geom = hv_union_geom(front, x_ref, y_ref)
        hv.append(float(fix_geom(hv_geom).area))

    hv = np.asarray(hv, dtype=float)

    if HV_MODE == "absolute":
        y = hv
    elif HV_MODE == "gain":
        y = hv - hv[0]
    elif HV_MODE == "normalized":
        denom = (hv[-1] - hv[0])
        y = (hv - hv[0]) / denom if abs(denom) > 1e-12 else np.zeros_like(hv)
    else:
        raise ValueError("HV_MODE must be one of: 'gain', 'absolute', 'normalized'")

    return batches, y, hv

def plot_hv_panel(ax, json_path: Path):
    batches, y, _ = compute_hv_series(json_path)

    main_color = "#FF42AD"

    ax.plot(batches, y, linewidth=LINE_LW, alpha=LINE_ALPHA, color=main_color, zorder=10)

    for b in batches:
        shiny_ball(ax, b, y[b - 1], s=SCATTER_S*1.2, color=main_color, lw=1.0, zorder=20)

    shiny_ball(ax, batches[0],  y[0],  s=HIGHLIGHT_S, color=main_color, lw=0.0, zorder=30)
    shiny_ball(ax, batches[-1], y[-1], s=HIGHLIGHT_S, color=main_color, lw=0.0, zorder=31)

    ax.set_xticks(batches)

    ax.tick_params(axis="both", which="both", labelsize=24)
    for t in ax.get_xticklabels() + ax.get_yticklabels():
        t.set_fontname("Arial")
        t.set_fontweight("bold")

    ax.margins(x=0.05, y=0.10)
    for side in ["left", "right", "top", "bottom"]:
        ax.spines[side].set_linewidth(FRAME_LW)

# -------------------------
# 2×4 figure
# -------------------------
ncol = len(systems)
fig_w = TOP_FIGSIZE_PER_PANEL[0] * ncol
fig_h = TOP_FIGSIZE_PER_PANEL[1] * 2

fig, axs = plt.subplots(
    2, ncol,
    figsize=(fig_w, fig_h),
    gridspec_kw={"height_ratios": [1.15, 1.00]}
)

# plot each column
for j, (p, sys_name) in enumerate(systems):
    ax_top = axs[0, j]
    ax_bot = axs[1, j]

    plot_system(ax_top, p)
    plot_hv_panel(ax_bot, p)

    # title only on the TOP row
    ax_top.set_title(sys_name, fontname="Arial", fontweight="bold", fontsize=30, pad=14)

    # remove top-row axis labels (use shared labels instead)
    ax_top.set_xlabel("")
    ax_top.set_ylabel("")

    # KEEP system name for bottom row (as xlabel)
    ax_bot.set_title(sys_name, fontname="Arial", fontweight="bold", fontsize=30, pad=14)
    ax_bot.set_xlabel("")
    ax_bot.set_ylabel("")


# leave room for shared labels
ROW_HPAD = 10.0  # increase to make the two rows further apart (try 2.0 ~ 5.0)
fig.tight_layout(rect=[0.11, 0.06, 0.99, 0.98], h_pad=ROW_HPAD)

# -------------------------
# shared labels (top row and bottom row separately)
# -------------------------
top_axes = list(axs[0, :].ravel())
bot_axes = list(axs[1, :].ravel())

bbox_top = mtransforms.Bbox.union([ax.get_position() for ax in top_axes])
bbox_bot = mtransforms.Bbox.union([ax.get_position() for ax in bot_axes])

pad = 0.02

# Top row shared labels
y_center_top = 0.5 * (bbox_top.y0 + bbox_top.y1)
x_center_top = 0.5 * (bbox_top.x0 + bbox_top.x1)
x_left_top   = bbox_top.x0

fig.text(
    x_left_top - 1.5*pad, y_center_top,
    "Separation Coefficient (@ -50 mA/cm²)",
    rotation="vertical", va="center", ha="center",
    fontname="Arial", fontweight="bold", fontsize=30
)

# put top-row xlabel between the two rows
y_between = 0.5 * (bbox_top.y0 + bbox_bot.y1)
fig.text(
    x_center_top, y_between + 1.2*pad,
    "Potential (@ -10 mA/cm², vs. RHE, V)",
    va="center", ha="center",
    fontname="Arial", fontweight="bold", fontsize=30
)

# Bottom row shared labels
y_center_bot = 0.5 * (bbox_bot.y0 + bbox_bot.y1)
x_center_bot = 0.5 * (bbox_bot.x0 + bbox_bot.x1)
x_left_bot   = bbox_bot.x0

y_label = {
    "gain": "Improvement of Hypervolume",
    "absolute": "Hypervolume",
    "normalized": "Normalized Hypervolume Gain (0–1)"
}[HV_MODE]

fig.text(
    x_left_bot - 1.5*pad, y_center_bot,
    y_label,
    rotation="vertical", va="center", ha="center",
    fontname="Arial", fontweight="bold", fontsize=30
)

fig.text(
    x_center_bot, 2.5*pad,
    "Batch Number",
    va="center", ha="center",
    fontname="Arial", fontweight="bold", fontsize=30
)

plt.show()
fig.savefig("eight_panels_2x4.png", dpi=600, bbox_inches="tight", pad_inches=0.02)
```

```python
import json
import numpy as np
import pandas as pd
from pathlib import Path
from shapely.geometry import Polygon, box
from shapely.ops import unary_union

# =========================
# keys
# =========================
voltage_key = "lsv_overpotential_-0.01"
sep_key     = "separation_coefficient"

systems = [
    (Path("initial_sampling_Co_Ni.json"), "Co-Ni"),
    (Path("initial_sampling_Co_Mo.json"), "Co-Mo"),
    (Path("initial_sampling_Fe_Ni.json"), "Fe-Ni"),
    (Path("initial_sampling_Co_V.json"),  "Co-V"),
]

# =========================
# batch0 data (Potential, Separation)
# 你给的 -101/-212 可能是 mV，这里自动转成 V
# =========================
BATCH0_BY_SYS = {
    "Co-Ni": (-0.185, 8.5),
    "Co-Mo": (-0.101,   9.8),   # -> -0.101
    "Fe-Ni": (-0.212,   9.0),   # -> -0.212
    "Co-V":  (-0.082, 9.3),
}

def _normalize_potential(x):
    x = float(x)
    return x / 1000.0 if abs(x) > 5 else x

# =========================
# helpers (与你原来一致)
# =========================
def dictcol_to_series(d):
    s = pd.Series(d)
    s.index = s.index.astype(int)
    return s.sort_index()

def load_system_df(json_path: Path):
    data = json.loads(json_path.read_text())

    v_ser = dictcol_to_series(data[voltage_key])
    s_ser = dictcol_to_series(data[sep_key])

    df = pd.DataFrame({
        "id": v_ser.index,
        "voltage": v_ser.values + 0.0592*14 + 0.098,      # Potential vs RHE
        "separation": s_ser.reindex(v_ser.index).values,
    }).dropna(subset=["voltage", "separation"]).reset_index(drop=True)

    df["id"] = df["id"].astype(int)
    df = df.sort_values("id").reset_index(drop=True)

    df["batch"] = (df["id"] // 4) + 1
    max_batch = int(df["batch"].max())

    # HV reference (你原来就是这个)
    x_ref = float(df["voltage"].min())
    y_ref = float(df["separation"].min())

    # 记录 reference 分别来自哪个点（不要求同一个点）
    ix = int(df["voltage"].idxmin())
    iy = int(df["separation"].idxmin())

    ref_from = {
        "min_voltage_point": df.loc[ix, ["id", "batch", "voltage", "separation"]].to_dict(),
        "min_separation_point": df.loc[iy, ["id", "batch", "voltage", "separation"]].to_dict(),
    }
    return df, max_batch, x_ref, y_ref, ref_from

def pareto_front_max(df_in: pd.DataFrame, x="voltage", y="separation"):
    s = df_in.sort_values([x, y], ascending=[False, False]).reset_index(drop=True)
    best_y = -np.inf
    keep = []
    for _, r in s.iterrows():
        if r[y] > best_y + 1e-12:
            keep.append(True)
            best_y = r[y]
        else:
            keep.append(False)
    front = s.loc[keep].copy()
    return front.sort_values(x, ascending=True).reset_index(drop=True)

def fix_geom(g):
    if g is None:
        return Polygon()
    if not g.is_valid:
        g = g.buffer(0)
    return g

def hv_union_geom(front: pd.DataFrame, x_ref: float, y_ref: float,
                  x="voltage", y="separation"):
    rects = []
    for _, r in front.iterrows():
        xi = float(r[x]); yi = float(r[y])
        if xi <= x_ref or yi <= y_ref:
            continue
        rects.append(box(x_ref, y_ref, xi, yi))
    if not rects:
        return Polygon()
    return fix_geom(unary_union(rects))

# =========================
# diagnostics printer
# =========================
def print_hv_diagnostics(json_path: Path, sys_name: str, force_max_batch: int = 7, print_rect_areas: bool = True):
    df_json, max_batch, x_ref, y_ref, ref_from = load_system_df(json_path)

    if sys_name not in BATCH0_BY_SYS:
        raise KeyError(f"Missing batch0 for {sys_name} in BATCH0_BY_SYS")

    # batch0 row（只用于 HV 序列诊断）
    x0, y0 = BATCH0_BY_SYS[sys_name]
    x0 = _normalize_potential(x0)
    y0 = float(y0)
    df0 = pd.DataFrame([{
        "id": -1,
        "voltage": x0,
        "separation": y0,
        "batch": 0
    }])

    df_all = pd.concat([df0, df_json], ignore_index=True)

    last_b = min(force_max_batch, max_batch)  # 通常 max_batch=7
    batches = list(range(0, last_b + 1))

    # ---- header
    print("\n" + "="*90)
    print(f"[{sys_name}]  file = {json_path}")
    print(f"HV reference (from JSON minima): x_ref={x_ref:.6f}, y_ref={y_ref:.6f}")
    print("reference 来源点（分别取 min voltage / min separation，不要求同一个点）：")
    print(f"  - min voltage point    : {ref_from['min_voltage_point']}")
    print(f"  - min separation point : {ref_from['min_separation_point']}")
    print(f"batch0 point used for HV series: id=-1, voltage={x0:.6f}, separation={y0:.6f}")
    print("-"*90)

    hv_list = []

    for b in batches:
        cum = df_all[df_all["batch"] <= b].copy()

        front = pareto_front_max(cum)
        # 真正贡献 HV 的 Pareto 点（必须超过 reference）
        contrib = front[(front["voltage"] > x_ref) & (front["separation"] > y_ref)].copy()

        hv_geom = hv_union_geom(front, x_ref, y_ref)
        hv_area = float(fix_geom(hv_geom).area)

        hv_list.append(hv_area)

        print(f"\nBatch {b}:")
        print(f"  cumulative points = {len(cum)}")
        print(f"  pareto points     = {len(front)}")
        print(f"  hv-contrib points = {len(contrib)}")
        print(f"  HV area           = {hv_area:.10f}")

        # 打印 Pareto 点列表
        show_cols = ["id", "batch", "voltage", "separation"]
        print("  Pareto front points (sorted by voltage asc):")
        print(front[show_cols].to_string(index=False))

        # 打印贡献点
        if len(contrib) > 0:
            print("  HV contributing points (voltage>x_ref & separation>y_ref):")
            print(contrib[show_cols].to_string(index=False))
        else:
            print("  HV contributing points: NONE")

        # （可选）每个贡献点的“单独矩形面积”（未去重）
        if print_rect_areas and len(contrib) > 0:
            rect_areas = (contrib["voltage"] - x_ref) * (contrib["separation"] - y_ref)
            tmp = contrib[show_cols].copy()
            tmp["rect_area_raw"] = rect_areas.values
            print("  Raw rectangle areas per contributing Pareto point (NOT union, for debugging):")
            print(tmp.to_string(index=False))

    hv = np.asarray(hv_list, dtype=float)
    improvement = hv - hv[0]
    improvement[0] = 0.0

    # 总结表
    print("\n" + "-"*90)
    summary = pd.DataFrame({
        "batch": batches,
        "HV": hv,
        "HV_improvement_vs_batch0": improvement
    })
    print("Summary (batch0→batch7):")
    print(summary.to_string(index=False, float_format=lambda x: f"{x:.10f}"))
    print("="*90 + "\n")

# =========================
# run for all systems
# =========================
for p, name in systems:
    print_hv_diagnostics(p, name, force_max_batch=7, print_rect_areas=True)
```

```python
import json
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
from shapely.geometry import Polygon, box
from shapely.ops import unary_union
import matplotlib.transforms as mtransforms

# -------------------------
# keys
# -------------------------
voltage_key = "lsv_overpotential_-0.01"
sep_key     = "separation_coefficient"

# -------------------------
# systems (4 columns)
# -------------------------
systems = [
    (Path("initial_sampling_Co_Ni.json"), "Co-Ni"),
    (Path("initial_sampling_Co_Mo.json"), "Co-Mo"),
    (Path("initial_sampling_Fe_Ni.json"), "Fe-Ni"),
    (Path("initial_sampling_Co_V.json"),  "Co-V"),
]

# ============================================================
# (1) Batch0 data (Potential, Separation)
# 你给的 Co-Mo / Fe-Ni 有可能是 -101/-212（mV），这里自动修正到 V
# ============================================================
BATCH0_BY_SYS = {
    "Co-Ni": (-0.185, 8.5),
    "Co-Mo": (-0.101,   9.8),   # 自动 -> -0.101
    "Fe-Ni": (-0.212,   9.0),   # 自动 -> -0.212
    "Co-V":  (-0.082, 9.3),
}

def _normalize_potential(x):
    x = float(x)
    return x / 1000.0 if abs(x) > 5 else x

# -------------------------
# TOP row style (Pareto + HV area)
# -------------------------
TOP_FIGSIZE_PER_PANEL = (8, 9)
S_SCATTER = 100
SCATTER_ALPHA = 0.95
AREA_ALPHA = 0.05
HV_LINE_ALPHA_EARLY = 0.80
HV_LINE_ALPHA_LAST  = 0.80
HV_LINE_LW = 2.0
PARETO_LINE_ALPHA_LAST = 0.80
PARETO_LINE_LW = 2.5
PARETO_MARKERSIZE = 20
PARETO_POINT_SIZE = 180
PARETO_POINT_ALPHA = 0.85
REVERSE_RAINBOW = True
FRAME_LW = 2.5

# -------------------------
# BOTTOM row style (HV vs batch)
# -------------------------
LINE_LW = 2.5
LINE_ALPHA = 0.85
SCATTER_S = 300
SCATTER_ALPHA = 0.95
HIGHLIGHT_S = 300

HV_MODE = "gain"  # 你要求 batch0 improvement=0，所以建议保持 gain

# -------------------------
# helpers
# -------------------------
def dictcol_to_series(d):
    s = pd.Series(d)
    s.index = s.index.astype(int)
    return s.sort_index()

def load_system_df(json_path: Path):
    data = json.loads(json_path.read_text())

    df = pd.DataFrame({
        "id": dictcol_to_series(data[voltage_key]).index,
        "voltage": dictcol_to_series(data[voltage_key]).values + 0.0592*14 + 0.098,
        "separation": dictcol_to_series(data[sep_key]).values,
    }).dropna(subset=["voltage", "separation"]).reset_index(drop=True)

    df["id"] = df["id"].astype(int)
    df = df.sort_values("id").reset_index(drop=True)

    # 原始 batch 仍然从 1 开始（保持 Pareto 图不变）
    df["batch"] = (df["id"] // 4) + 1
    max_batch = int(df["batch"].max())

    # (2) reference：每个 JSON 各自用 min(Potential) & min(Separation) 组成参考点（你原来就这么做）
    x_ref = float(df["voltage"].min())
    y_ref = float(df["separation"].min())
    return df, max_batch, x_ref, y_ref

def pareto_front_max(df_in: pd.DataFrame, x="voltage", y="separation"):
    s = df_in.sort_values([x, y], ascending=[False, False]).reset_index(drop=True)
    best_y = -np.inf
    keep = []
    for _, r in s.iterrows():
        if r[y] > best_y + 1e-12:
            keep.append(True)
            best_y = r[y]
        else:
            keep.append(False)
    front = s.loc[keep].copy()
    return front.sort_values(x, ascending=True).reset_index(drop=True)

def fix_geom(g):
    if g is None:
        return Polygon()
    if not g.is_valid:
        g = g.buffer(0)
    return g

def iter_polygons(g):
    if g is None or g.is_empty:
        return
    gt = g.geom_type
    if gt == "Polygon":
        yield g
    elif gt == "MultiPolygon":
        yield from g.geoms
    elif gt == "GeometryCollection":
        for gg in g.geoms:
            yield from iter_polygons(gg)

def draw_geom_boundary(ax, geom, color, alpha=0.3, lw=2.0):
    for p in iter_polygons(geom):
        x, y = p.exterior.xy
        ax.plot(x, y, color=color, alpha=alpha, linewidth=lw)

def fill_geom(ax, geom, color, alpha=0.05):
    for p in iter_polygons(geom):
        x, y = p.exterior.xy
        ax.fill(x, y, color=color, alpha=alpha, linewidth=0)

def hv_union_geom(front: pd.DataFrame, x_ref: float, y_ref: float,
                  x="voltage", y="separation"):
    rects = []
    for _, r in front.iterrows():
        xi = float(r[x]); yi = float(r[y])
        if xi <= x_ref or yi <= y_ref:
            continue
        rects.append(box(x_ref, y_ref, xi, yi))
    if not rects:
        return Polygon()
    return fix_geom(unary_union(rects))

def shiny_ball(ax, x, y, s=140, color=None, edgecolor="black", lw=1.0,
               highlight_alpha=0.35, zorder=20):
    if color is None:
        color = plt.rcParams["axes.prop_cycle"].by_key()["color"][0]

    ax.scatter([x], [y], s=s, color=color, edgecolors=edgecolor, linewidths=lw, zorder=zorder)

    fig = ax.figure
    offset = mtransforms.ScaledTranslation(-2/72., 2/72., fig.dpi_scale_trans)
    ax.scatter([x], [y],
               s=s * 0.22,
               color="white",
               alpha=highlight_alpha,
               linewidths=0,
               transform=ax.transData + offset,
               zorder=zorder + 10)

# -------------------------
# TOP row plotting
# (4) batch0 只画灰点，不参与 Pareto/HV 的计算（保证“Pareto front 图不变”）
# -------------------------
def plot_system(ax, json_path: Path, sys_name: str):
    df, max_batch, x_ref, y_ref = load_system_df(json_path)

    # 先叠加 batch0 灰点
    if sys_name in BATCH0_BY_SYS:
        x0, y0 = BATCH0_BY_SYS[sys_name]
        x0 = _normalize_potential(x0)
        ax.scatter([x0], [y0],
                   s=S_SCATTER*1.2, color="lightgray",
                   edgecolors="black", linewidths=0.8, zorder=200)

    # 下面完全沿用你原来的 Pareto/HV 逻辑（不变）
    fronts, hv_geoms = {}, {}
    for b in range(1, max_batch + 1):
        cum = df[df["batch"] <= b].copy()
        fronts[b] = pareto_front_max(cum)
        hv_geoms[b] = hv_union_geom(fronts[b], x_ref, y_ref)

    cmap = plt.cm.rainbow
    colors = [cmap(i / max(1, (max_batch - 1))) for i in range(max_batch)]
    if REVERSE_RAINBOW:
        colors = colors[::-1]

    for b in range(1, max_batch + 1):
        c = colors[b - 1]
        dfb = df[df["batch"] == b]
        ax.scatter(dfb["voltage"], dfb["separation"],
                   s=S_SCATTER, color=c, alpha=SCATTER_ALPHA,
                   edgecolors="black", linewidths=0.8, zorder=110)

    c1 = colors[0]
    cl = colors[max_batch - 1]

    hv1 = fix_geom(hv_geoms[1])
    hvl = fix_geom(hv_geoms[max_batch])

    fill_geom(ax, hv1, c1, alpha=AREA_ALPHA)
    if max_batch > 1:
        delta_last = fix_geom(hvl.difference(hv1))
        fill_geom(ax, delta_last, cl, alpha=AREA_ALPHA)

    draw_geom_boundary(ax, hv1, c1, alpha=HV_LINE_ALPHA_EARLY, lw=HV_LINE_LW)
    if max_batch > 1:
        draw_geom_boundary(ax, hvl, cl, alpha=HV_LINE_ALPHA_LAST, lw=HV_LINE_LW)

    f1 = fronts[1]
    ax.scatter(f1["voltage"], f1["separation"],
               s=PARETO_POINT_SIZE, color=c1,
               edgecolors="black", linewidths=0.8,
               alpha=PARETO_POINT_ALPHA, zorder=120)
    ax.plot(f1["voltage"], f1["separation"],
            linewidth=PARETO_LINE_LW, markersize=PARETO_MARKERSIZE,
            color=c1, alpha=PARETO_LINE_ALPHA_LAST, linestyle="--")

    f_last = fronts[max_batch]
    ax.scatter(f_last["voltage"], f_last["separation"],
               s=PARETO_POINT_SIZE, color=cl,
               edgecolors="black", linewidths=0.8,
               alpha=PARETO_POINT_ALPHA, zorder=130)
    ax.plot(f_last["voltage"], f_last["separation"],
            linewidth=PARETO_LINE_LW, markersize=PARETO_MARKERSIZE,
            color=cl, alpha=PARETO_LINE_ALPHA_LAST, linestyle="--")

    from matplotlib.ticker import MultipleLocator
    ax.yaxis.set_major_locator(MultipleLocator(1))
    ax.xaxis.set_major_locator(MultipleLocator(0.05))
    ax.tick_params(axis="both", which="both", labelsize=24)
    for t in ax.get_xticklabels() + ax.get_yticklabels():
        t.set_fontname("Arial")
        t.set_fontweight("bold")
    for side in ["left", "right", "top", "bottom"]:
        ax.spines[side].set_linewidth(FRAME_LW)

# -------------------------
# BOTTOM row plotting
# (1)(2)(3) 加入 batch0；HV reference = JSON minima；improvement 从 batch0 到 batch7，batch0=0
# -------------------------
def compute_hv_series(json_path: Path, sys_name: str):
    df_json, max_batch, x_ref, y_ref = load_system_df(json_path)

    # 构造 batch0 数据（只用于 HV improvement 曲线）
    if sys_name not in BATCH0_BY_SYS:
        raise KeyError(f"Missing batch0 definition for {sys_name} in BATCH0_BY_SYS")

    x0, y0 = BATCH0_BY_SYS[sys_name]
    x0 = _normalize_potential(x0)
    y0 = float(y0)

    df0 = pd.DataFrame({
        "id": [-1],
        "voltage": [x0],
        "separation": [y0],
        "batch": [0],
    })

    df_all = pd.concat([df0, df_json], ignore_index=True)

    # batch 从 0..max_batch（通常 max_batch=7）
    batches = np.arange(0, max_batch + 1, dtype=int)

    hv = []
    for b in batches:
        cum = df_all[df_all["batch"] <= b].copy()
        front = pareto_front_max(cum)
        hv_geom = hv_union_geom(front, x_ref, y_ref)  # reference = JSON minima
        hv.append(float(fix_geom(hv_geom).area))

    hv = np.asarray(hv, dtype=float)

    if HV_MODE == "absolute":
        y = hv
    elif HV_MODE == "gain":
        # (3) batch0 improvement = 0
        y = hv - hv[0]
        y[0] = 0.0
    elif HV_MODE == "normalized":
        denom = (hv[-1] - hv[0])
        y = (hv - hv[0]) / denom if abs(denom) > 1e-12 else np.zeros_like(hv)
        y[0] = 0.0
    else:
        raise ValueError("HV_MODE must be one of: 'gain', 'absolute', 'normalized'")

    return batches, y, hv

def plot_hv_panel(ax, json_path: Path, sys_name: str):
    batches, y, _ = compute_hv_series(json_path, sys_name)

    main_color = "#FF42AD"
    ax.plot(batches, y, linewidth=LINE_LW, alpha=LINE_ALPHA, color=main_color, zorder=10)

    for i, b in enumerate(batches):
        shiny_ball(ax, b, y[i], s=SCATTER_S*1.2, color=main_color, lw=1.0, zorder=20)

    shiny_ball(ax, batches[0],  y[0],  s=HIGHLIGHT_S, color=main_color, lw=0.0, zorder=30)
    shiny_ball(ax, batches[-1], y[-1], s=HIGHLIGHT_S, color=main_color, lw=0.0, zorder=31)

    ax.set_xticks(batches)  # ✅ 包含 0

    ax.tick_params(axis="both", which="both", labelsize=24)
    for t in ax.get_xticklabels() + ax.get_yticklabels():
        t.set_fontname("Arial")
        t.set_fontweight("bold")

    ax.margins(x=0.05, y=0.10)
    for side in ["left", "right", "top", "bottom"]:
        ax.spines[side].set_linewidth(FRAME_LW)

# -------------------------
# 2×4 figure
# -------------------------
ncol = len(systems)
fig_w = TOP_FIGSIZE_PER_PANEL[0] * ncol
fig_h = TOP_FIGSIZE_PER_PANEL[1] * 2

fig, axs = plt.subplots(
    2, ncol,
    figsize=(fig_w, fig_h),
    gridspec_kw={"height_ratios": [1.15, 1.00]}
)

for j, (p, sys_name) in enumerate(systems):
    ax_top = axs[0, j]
    ax_bot = axs[1, j]

    plot_system(ax_top, p, sys_name)
    plot_hv_panel(ax_bot, p, sys_name)

    ax_top.set_title(sys_name, fontname="Arial", fontweight="bold", fontsize=30, pad=14)
    ax_top.set_xlabel("")
    ax_top.set_ylabel("")

    ax_bot.set_title(sys_name, fontname="Arial", fontweight="bold", fontsize=30, pad=14)
    ax_bot.set_xlabel("")
    ax_bot.set_ylabel("")

ROW_HPAD = 10.0
fig.tight_layout(rect=[0.11, 0.06, 0.99, 0.98], h_pad=ROW_HPAD)

top_axes = list(axs[0, :].ravel())
bot_axes = list(axs[1, :].ravel())
bbox_top = mtransforms.Bbox.union([ax.get_position() for ax in top_axes])
bbox_bot = mtransforms.Bbox.union([ax.get_position() for ax in bot_axes])

pad = 0.02
y_center_top = 0.5 * (bbox_top.y0 + bbox_top.y1)
x_center_top = 0.5 * (bbox_top.x0 + bbox_top.x1)
x_left_top   = bbox_top.x0

fig.text(
    x_left_top - 1.5*pad, y_center_top,
    "Separation Coefficient (@ -50 mA/cm²)",
    rotation="vertical", va="center", ha="center",
    fontname="Arial", fontweight="bold", fontsize=30
)

y_between = 0.5 * (bbox_top.y0 + bbox_bot.y1)
fig.text(
    x_center_top, y_between + 1.2*pad,
    "Potential (@ -10 mA/cm², vs. RHE, V)",
    va="center", ha="center",
    fontname="Arial", fontweight="bold", fontsize=30
)

y_center_bot = 0.5 * (bbox_bot.y0 + bbox_bot.y1)
x_center_bot = 0.5 * (bbox_bot.x0 + bbox_bot.x1)
x_left_bot   = bbox_bot.x0

y_label = {
    "gain": "Improvement of Hypervolume",
    "absolute": "Hypervolume",
    "normalized": "Normalized Hypervolume Gain (0–1)"
}[HV_MODE]

fig.text(
    x_left_bot - 1.5*pad, y_center_bot,
    y_label,
    rotation="vertical", va="center", ha="center",
    fontname="Arial", fontweight="bold", fontsize=30
)

fig.text(
    x_center_bot, 2.5*pad,
    "Batch Number",
    va="center", ha="center",
    fontname="Arial", fontweight="bold", fontsize=30
)

plt.show()
fig.savefig("eight_panels_2x4-batch1.png", dpi=300, bbox_inches="tight", pad_inches=0.02)
```

```python
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D
from matplotlib.legend_handler import HandlerBase
import matplotlib as mpl

# ============ 字体：Arial ============
mpl.rcParams["font.family"] = "Arial"

# ============ 批次颜色（Batch0~Batch7）===========
# 你如果要和主图完全一致，把这里换成主图那套颜色
batch_colors = {
    0: "#bfbfbf",  # Batch0（你可按需改）
    1: "#ff2d2d",  # red
    2: "#ff8c42",  # orange
    3: "#d6df6a",  # yellow-green
    4: "#66e6a1",  # green
    5: "#44e6e6",  # cyan
    6: "#4a7dff",  # blue
    7: "#8a2be2",  # purple
}

# ============ 自定义：两条虚线“竖着叠放”的 legend handler ============
class ParetoHandle:
    """占位用的 handle 类型"""

class HandlerTwoDashedLines(HandlerBase):
    def __init__(self, colors, lw=2.0, ls="--", **kwargs):
        super().__init__(**kwargs)
        self.colors = colors  # (top, bottom)
        self.lw = lw
        self.ls = ls

    def create_artists(self, legend, orig_handle, x0, y0, width, height, fontsize, trans):
        x1 = x0 + width
        y_top = y0 + 0.72 * height
        y_bot = y0 + 0.28 * height

        line_top = Line2D([x0, x1], [y_top, y_top],
                          linestyle=self.ls, linewidth=self.lw,
                          color=self.colors[0], transform=trans)
        line_bot = Line2D([x0, x1], [y_bot, y_bot],
                          linestyle=self.ls, linewidth=self.lw,
                          color=self.colors[1], transform=trans)
        return [line_top, line_bot]

# ============ 生成 legend ============
handles, labels = [], []

for b in range(0, 8):  # ✅ Batch0~Batch7
    h = Line2D([0], [0],
               marker="o", linestyle="None",
               markerfacecolor=batch_colors[b],
               markeredgecolor="black",
               markeredgewidth=1.2,
               markersize=10)
    handles.append(h)
    labels.append(f"Batch {b}")

# Pareto Front：上红(批次1色) 下紫(批次7色)
pareto = ParetoHandle()
handles.append(pareto)
labels.append("Pareto Front")

# ============ 画布只放 legend ============
fig = plt.figure(figsize=(20, 1.2), dpi=300)  # 稍微加宽，容纳 Batch0
ax = fig.add_axes([0, 0, 1, 1])
ax.axis("off")

leg = ax.legend(
    handles, labels,
    loc="center",
    ncol=len(labels),
    frameon=True,
    fancybox=False,
    edgecolor="black",
    framealpha=1.0,
    borderpad=0.4,
    handletextpad=0.6,
    columnspacing=1.1,
    prop={"family": "Arial", "weight": "bold", "size": 14},
    handler_map={
        ParetoHandle: HandlerTwoDashedLines(
            colors=(batch_colors[1], batch_colors[7]),  # 上红下紫
            lw=2.0,  # 不加粗
            ls="--"
        )
    }
)

leg.get_frame().set_linewidth(1.5)

out_png = "custom_legend_batch0_7_arial.png"
out_svg = "custom_legend_batch0_7_arial.svg"
fig.savefig(out_png, bbox_inches="tight", pad_inches=0.05)
fig.savefig(out_svg, bbox_inches="tight", pad_inches=0.05)
plt.close(fig)

print("Saved:", out_png, out_svg)
```

```python
import json
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
from shapely.geometry import Polygon, box
from shapely.ops import unary_union
import matplotlib.transforms as mtransforms

# -------------------------
# keys
# -------------------------
voltage_key = "lsv_overpotential_-0.01"
sep_key     = "separation_coefficient"

# -------------------------
# systems (4 columns)
# -------------------------
systems = [
    (Path("initial_sampling_Co_Ni.json"), "Co-Ni"),
    (Path("initial_sampling_Co_Mo.json"), "Co-Mo"),
    (Path("initial_sampling_Fe_Ni.json"), "Fe-Ni"),
    (Path("initial_sampling_Co_V.json"),  "Co-V"),
]

# -------------------------
# Batch0 point for each system (Potential, Separation)
# -------------------------
BATCH0_BY_SYS = {
    "Co-Ni": (-0.185, 8.5),
    "Co-Mo": (-0.101, 9.8),   # if you ever pass -101, will be auto /1000
    "Fe-Ni": (-0.212, 9.0),
    "Co-V":  (-0.082, 9.3),
}

def _normalize_potential(x):
    x = float(x)
    return x / 1000.0 if abs(x) > 5 else x

# -------------------------
# TOP row style (Pareto + HV area)
# -------------------------
TOP_FIGSIZE_PER_PANEL = (8, 9)  # (w,h) per panel in inches
S_SCATTER = 100
SCATTER_ALPHA = 0.95
AREA_ALPHA = 0.05
HV_LINE_ALPHA_EARLY = 0.80
HV_LINE_ALPHA_LAST  = 0.80
HV_LINE_LW = 2.0
PARETO_LINE_ALPHA_LAST = 0.80
PARETO_LINE_LW = 2.5
PARETO_MARKERSIZE = 20
PARETO_POINT_SIZE = 180
PARETO_POINT_ALPHA = 0.85
REVERSE_RAINBOW = True
FRAME_LW = 2.5

# -------------------------
# BOTTOM row style (HV vs batch)
# -------------------------
LINE_LW = 2.5
LINE_ALPHA = 0.85
SCATTER_S = 300
SCATTER_ALPHA = 0.95
HIGHLIGHT_S = 300

# choose what to plot:
#   "gain"       : HV(b) - HV(0)   (starts at 0)
#   "absolute"   : HV(b)
#   "normalized" : (HV(b)-HV(0)) / (HV(last)-HV(0))   (0→1)
HV_MODE = "gain"

# -------------------------
# helpers
# -------------------------
def dictcol_to_series(d):
    s = pd.Series(d)
    s.index = s.index.astype(int)
    return s.sort_index()

def load_system_df(json_path: Path):
    data = json.loads(json_path.read_text())

    df = pd.DataFrame({
        "id": dictcol_to_series(data[voltage_key]).index,
        "voltage": dictcol_to_series(data[voltage_key]).values + 0.0592*14 + 0.098,
        "separation": dictcol_to_series(data[sep_key]).values,
    }).dropna(subset=["voltage", "separation"]).reset_index(drop=True)

    df["id"] = df["id"].astype(int)
    df = df.sort_values("id").reset_index(drop=True)
    df["batch"] = (df["id"] // 4) + 1
    max_batch = int(df["batch"].max())

    # HV reference = dataset minima (as you used before)
    x_ref = float(df["voltage"].min())
    y_ref = float(df["separation"].min())
    return df, max_batch, x_ref, y_ref

def pareto_front_max(df_in: pd.DataFrame, x="voltage", y="separation"):
    s = df_in.sort_values([x, y], ascending=[False, False]).reset_index(drop=True)
    best_y = -np.inf
    keep = []
    for _, r in s.iterrows():
        if r[y] > best_y + 1e-12:
            keep.append(True)
            best_y = r[y]
        else:
            keep.append(False)
    front = s.loc[keep].copy()
    return front.sort_values(x, ascending=True).reset_index(drop=True)

def fix_geom(g):
    if g is None:
        return Polygon()
    if not g.is_valid:
        g = g.buffer(0)
    return g

def iter_polygons(g):
    if g is None or g.is_empty:
        return
    gt = g.geom_type
    if gt == "Polygon":
        yield g
    elif gt == "MultiPolygon":
        yield from g.geoms
    elif gt == "GeometryCollection":
        for gg in g.geoms:
            yield from iter_polygons(gg)

def draw_geom_boundary(ax, geom, color, alpha=0.3, lw=2.0):
    for p in iter_polygons(geom):
        x, y = p.exterior.xy
        ax.plot(x, y, color=color, alpha=alpha, linewidth=lw)

def fill_geom(ax, geom, color, alpha=0.05):
    for p in iter_polygons(geom):
        x, y = p.exterior.xy
        ax.fill(x, y, color=color, alpha=alpha, linewidth=0)

def hv_union_geom(front: pd.DataFrame, x_ref: float, y_ref: float,
                  x="voltage", y="separation"):
    rects = []
    for _, r in front.iterrows():
        xi = float(r[x]); yi = float(r[y])
        if xi <= x_ref or yi <= y_ref:
            continue
        rects.append(box(x_ref, y_ref, xi, yi))
    if not rects:
        return Polygon()
    return fix_geom(unary_union(rects))

def shiny_ball(ax, x, y, s=140, color=None, edgecolor="black", lw=1.0,
               highlight_alpha=0.35, zorder=20):
    if color is None:
        color = plt.rcParams["axes.prop_cycle"].by_key()["color"][0]

    ax.scatter([x], [y], s=s, color=color, edgecolors=edgecolor, linewidths=lw, zorder=zorder)

    fig = ax.figure
    offset = mtransforms.ScaledTranslation(-2/72., 2/72., fig.dpi_scale_trans)
    ax.scatter([x], [y],
               s=s * 0.22,
               color="white",
               alpha=highlight_alpha,
               linewidths=0,
               transform=ax.transData + offset,
               zorder=zorder + 10)

# -------------------------
# TOP row plotting
# 改动点：把 batch0 真正加入 fronts/hv 的计算；只画 batch0 & batch7 的 Pareto front
# 以及 ✅ 修正颜色索引：colors[b] 对应 batch=b
# -------------------------
def plot_system(ax, json_path: Path, sys_name: str):
    df_json, max_batch, x_ref, y_ref = load_system_df(json_path)

    if sys_name not in BATCH0_BY_SYS:
        raise KeyError(f"Missing batch0 definition for {sys_name} in BATCH0_BY_SYS")

    # ---- batch0 point (参与计算 + 灰点展示)
    x0, y0 = BATCH0_BY_SYS[sys_name]
    x0 = _normalize_potential(x0)
    y0 = float(y0)

    df0 = pd.DataFrame({
        "id": [-1],
        "voltage": [x0],
        "separation": [y0],
        "batch": [0],
    })

    # df_all 用于“计算 Pareto/HV”
    df_all = pd.concat([df0, df_json], ignore_index=True)

    # ---- compute fronts & hv for b=0..max_batch
    fronts, hv_geoms = {}, {}
    for b in range(0, max_batch + 1):
        cum = df_all[df_all["batch"] <= b].copy()
        fronts[b] = pareto_front_max(cum)
        hv_geoms[b] = hv_union_geom(fronts[b], x_ref, y_ref)

    # ---- colors (batch0..batch7 一一对应)
    cmap = plt.cm.rainbow
    colors = [cmap(i / max(1, max_batch)) for i in range(max_batch + 1)]  # index = batch
    if REVERSE_RAINBOW:
        colors = colors[::-1]

    early_color = colors[0]          # batch0
    last_color  = colors[max_batch]  # batch7

    # ---- scatter: batch0 灰点（展示）
    ax.scatter([x0], [y0],
               s=S_SCATTER*1.2, color="lightgray",
               edgecolors="black", linewidths=0.8, zorder=200)

    # ---- scatter: batches 1..max_batch（✅ 正确：用 colors[b]）
    for b in range(1, max_batch + 1):
        c = colors[b]
        dfb = df_json[df_json["batch"] == b]
        ax.scatter(dfb["voltage"], dfb["separation"],
                   s=S_SCATTER, color=c, alpha=SCATTER_ALPHA,
                   edgecolors="black", linewidths=0.8, zorder=110)

    # ---- HV fill/boundary: 早期用 batch0，最后用 batch7
    hv0 = fix_geom(hv_geoms[0])
    hvl = fix_geom(hv_geoms[max_batch])

    fill_geom(ax, hv0, early_color, alpha=AREA_ALPHA)
    if max_batch > 0:
        delta_last = fix_geom(hvl.difference(hv0))
        fill_geom(ax, delta_last, last_color, alpha=AREA_ALPHA)

    draw_geom_boundary(ax, hv0, early_color, alpha=HV_LINE_ALPHA_EARLY, lw=HV_LINE_LW)
    draw_geom_boundary(ax, hvl, last_color,  alpha=HV_LINE_ALPHA_LAST,  lw=HV_LINE_LW)

    # ---- Pareto front: 只画 batch0 & batch7
    f0 = fronts[0]
    ax.scatter(f0["voltage"], f0["separation"],
               s=PARETO_POINT_SIZE, color=early_color,
               edgecolors="black", linewidths=0.8,
               alpha=1.0, zorder=220)

    f_last = fronts[max_batch]
    ax.scatter(f_last["voltage"], f_last["separation"],
               s=PARETO_POINT_SIZE, color=last_color,
               edgecolors="black", linewidths=0.8,
               alpha=PARETO_POINT_ALPHA, zorder=130)
    ax.plot(f_last["voltage"], f_last["separation"],
            linewidth=PARETO_LINE_LW, markersize=PARETO_MARKERSIZE,
            color=last_color, alpha=PARETO_LINE_ALPHA_LAST, linestyle="--")

    # ---- ticks + frame (不变)
    from matplotlib.ticker import MultipleLocator
    ax.yaxis.set_major_locator(MultipleLocator(1))
    ax.xaxis.set_major_locator(MultipleLocator(0.05))
    ax.tick_params(axis="both", which="both", labelsize=24)
    for t in ax.get_xticklabels() + ax.get_yticklabels():
        t.set_fontname("Arial")
        t.set_fontweight("bold")
    for side in ["left", "right", "top", "bottom"]:
        ax.spines[side].set_linewidth(FRAME_LW)

# -------------------------
# BOTTOM row plotting (从 batch0 到 batch7，improvement 从 0 开始)
# -------------------------
def compute_hv_series(json_path: Path, sys_name: str):
    df_json, max_batch, x_ref, y_ref = load_system_df(json_path)

    if sys_name not in BATCH0_BY_SYS:
        raise KeyError(f"Missing batch0 definition for {sys_name} in BATCH0_BY_SYS")

    x0, y0 = BATCH0_BY_SYS[sys_name]
    x0 = _normalize_potential(x0)
    y0 = float(y0)

    df0 = pd.DataFrame({
        "id": [-1],
        "voltage": [x0],
        "separation": [y0],
        "batch": [0],
    })

    df_all = pd.concat([df0, df_json], ignore_index=True)

    batches = np.arange(0, max_batch + 1, dtype=int)

    hv = []
    for b in batches:
        cum = df_all[df_all["batch"] <= b].copy()
        front = pareto_front_max(cum)
        hv_geom = hv_union_geom(front, x_ref, y_ref)
        hv.append(float(fix_geom(hv_geom).area))

    hv = np.asarray(hv, dtype=float)

    if HV_MODE == "absolute":
        y = hv
    elif HV_MODE == "gain":
        y = hv - hv[0]
        y[0] = 0.0
    elif HV_MODE == "normalized":
        denom = (hv[-1] - hv[0])
        y = (hv - hv[0]) / denom if abs(denom) > 1e-12 else np.zeros_like(hv)
        y[0] = 0.0
    else:
        raise ValueError("HV_MODE must be one of: 'gain', 'absolute', 'normalized'")

    return batches, y, hv

def plot_hv_panel(ax, json_path: Path, sys_name: str):
    batches, y, _ = compute_hv_series(json_path, sys_name)

    main_color = "#FF42AD"
    ax.plot(batches, y, linewidth=LINE_LW, alpha=LINE_ALPHA, color=main_color, zorder=10)

    for i, b in enumerate(batches):
        shiny_ball(ax, b, y[i], s=SCATTER_S*1.2, color=main_color, lw=1.0, zorder=20)

    shiny_ball(ax, batches[0],  y[0],  s=HIGHLIGHT_S, color=main_color, lw=0.0, zorder=30)
    shiny_ball(ax, batches[-1], y[-1], s=HIGHLIGHT_S, color=main_color, lw=0.0, zorder=31)

    ax.set_xticks(batches)

    ax.tick_params(axis="both", which="both", labelsize=24)
    for t in ax.get_xticklabels() + ax.get_yticklabels():
        t.set_fontname("Arial")
        t.set_fontweight("bold")

    ax.margins(x=0.05, y=0.10)
    for side in ["left", "right", "top", "bottom"]:
        ax.spines[side].set_linewidth(FRAME_LW)

# -------------------------
# 2×4 figure
# -------------------------
ncol = len(systems)
fig_w = TOP_FIGSIZE_PER_PANEL[0] * ncol
fig_h = TOP_FIGSIZE_PER_PANEL[1] * 2

fig, axs = plt.subplots(
    2, ncol,
    figsize=(fig_w, fig_h),
    gridspec_kw={"height_ratios": [1.15, 1.00]}
)

for j, (p, sys_name) in enumerate(systems):
    ax_top = axs[0, j]
    ax_bot = axs[1, j]

    plot_system(ax_top, p, sys_name)
    plot_hv_panel(ax_bot, p, sys_name)

    ax_top.set_title(sys_name, fontname="Arial", fontweight="bold", fontsize=30, pad=14)
    ax_top.set_xlabel("")
    ax_top.set_ylabel("")

    ax_bot.set_title(sys_name, fontname="Arial", fontweight="bold", fontsize=30, pad=14)
    ax_bot.set_xlabel("")
    ax_bot.set_ylabel("")

ROW_HPAD = 10.0
fig.tight_layout(rect=[0.11, 0.06, 0.99, 0.98], h_pad=ROW_HPAD)

top_axes = list(axs[0, :].ravel())
bot_axes = list(axs[1, :].ravel())
bbox_top = mtransforms.Bbox.union([ax.get_position() for ax in top_axes])
bbox_bot = mtransforms.Bbox.union([ax.get_position() for ax in bot_axes])

pad = 0.02
y_center_top = 0.5 * (bbox_top.y0 + bbox_top.y1)
x_center_top = 0.5 * (bbox_top.x0 + bbox_top.x1)
x_left_top   = bbox_top.x0

fig.text(
    x_left_top - 1.5*pad, y_center_top,
    "Separation Coefficient (@ -50 mA/cm²)",
    rotation="vertical", va="center", ha="center",
    fontname="Arial", fontweight="bold", fontsize=30
)

y_between = 0.5 * (bbox_top.y0 + bbox_bot.y1)
fig.text(
    x_center_top, y_between + 1.2*pad,
    "Potential (@ -10 mA/cm², vs. RHE, V)",
    va="center", ha="center",
    fontname="Arial", fontweight="bold", fontsize=30
)

y_center_bot = 0.5 * (bbox_bot.y0 + bbox_bot.y1)
x_center_bot = 0.5 * (bbox_bot.x0 + bbox_bot.x1)
x_left_bot   = bbox_bot.x0

y_label = {
    "gain": "Improvement of Hypervolume",
    "absolute": "Hypervolume",
    "normalized": "Normalized Hypervolume Gain (0–1)"
}[HV_MODE]

fig.text(
    x_left_bot - 1.5*pad, y_center_bot,
    y_label,
    rotation="vertical", va="center", ha="center",
    fontname="Arial", fontweight="bold", fontsize=30
)

fig.text(
    x_center_bot, 2.5*pad,
    "Batch Number",
    va="center", ha="center",
    fontname="Arial", fontweight="bold", fontsize=30
)

plt.show()
fig.savefig("eight_panels_2x4.png", dpi=600, bbox_inches="tight", pad_inches=0.02)

```

```python
import json
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
from shapely.geometry import Polygon, box
from shapely.ops import unary_union
import matplotlib.transforms as mtransforms

# -------------------------
# keys
# -------------------------
voltage_key = "lsv_overpotential_-0.01"
sep_key     = "separation_coefficient"

# -------------------------
# systems (4 columns)
# -------------------------
systems = [
    (Path("initial_sampling_Co_Ni.json"), "Co-Ni"),
    (Path("initial_sampling_Co_Mo.json"), "Co-Mo"),
    (Path("initial_sampling_Fe_Ni.json"), "Fe-Ni"),
    (Path("initial_sampling_Co_V.json"),  "Co-V"),
]

# -------------------------
# Batch0 point for each system (Potential, Separation)
# -------------------------
BATCH0_BY_SYS = {
    "Co-Ni": (-0.185, 8.5),
    "Co-Mo": (-0.101, 9.8),
    "Fe-Ni": (-0.212, 9.0),
    "Co-V":  (-0.082, 9.3),
}

def _normalize_potential(x):
    x = float(x)
    return x / 1000.0 if abs(x) > 5 else x

# -------------------------
# TOP row style (Pareto + HV area)
# -------------------------
TOP_FIGSIZE_PER_PANEL = (8, 9)
S_SCATTER = 100
SCATTER_ALPHA = 0.95
AREA_ALPHA = 0.05
HV_LINE_ALPHA_EARLY = 0.80
HV_LINE_ALPHA_LAST  = 0.80
HV_LINE_LW = 2.0
PARETO_LINE_ALPHA_LAST = 0.80
PARETO_LINE_LW = 2.5
PARETO_MARKERSIZE = 20
PARETO_POINT_SIZE = 180
PARETO_POINT_ALPHA = 0.85
REVERSE_RAINBOW = True
FRAME_LW = 2.5

# -------------------------
# BOTTOM row style (HV vs batch)
# -------------------------
LINE_LW = 2.5
LINE_ALPHA = 0.85
SCATTER_S = 300
SCATTER_ALPHA = 0.95
HIGHLIGHT_S = 300

HV_MODE = "gain"

# -------------------------
# helpers
# -------------------------
def dictcol_to_series(d):
    s = pd.Series(d)
    s.index = s.index.astype(int)
    return s.sort_index()

def load_system_df(json_path: Path):
    data = json.loads(json_path.read_text())

    df = pd.DataFrame({
        "id": dictcol_to_series(data[voltage_key]).index,
        "voltage": dictcol_to_series(data[voltage_key]).values + 0.0592*14 + 0.098,
        "separation": dictcol_to_series(data[sep_key]).values,
    }).dropna(subset=["voltage", "separation"]).reset_index(drop=True)

    df["id"] = df["id"].astype(int)
    df = df.sort_values("id").reset_index(drop=True)
    df["batch"] = (df["id"] // 4) + 1
    max_batch = int(df["batch"].max())

    x_ref = float(df["voltage"].min())
    y_ref = float(df["separation"].min())
    return df, max_batch, x_ref, y_ref

def pareto_front_max(df_in: pd.DataFrame, x="voltage", y="separation"):
    s = df_in.sort_values([x, y], ascending=[False, False]).reset_index(drop=True)
    best_y = -np.inf
    keep = []
    for _, r in s.iterrows():
        if r[y] > best_y + 1e-12:
            keep.append(True)
            best_y = r[y]
        else:
            keep.append(False)
    front = s.loc[keep].copy()
    return front.sort_values(x, ascending=True).reset_index(drop=True)

def fix_geom(g):
    if g is None:
        return Polygon()
    if not g.is_valid:
        g = g.buffer(0)
    return g

def iter_polygons(g):
    if g is None or g.is_empty:
        return
    gt = g.geom_type
    if gt == "Polygon":
        yield g
    elif gt == "MultiPolygon":
        yield from g.geoms
    elif gt == "GeometryCollection":
        for gg in g.geoms:
            yield from iter_polygons(gg)

def draw_geom_boundary(ax, geom, color, alpha=0.3, lw=2.0):
    for p in iter_polygons(geom):
        x, y = p.exterior.xy
        ax.plot(x, y, color=color, alpha=alpha, linewidth=lw)

def fill_geom(ax, geom, color, alpha=0.05):
    for p in iter_polygons(geom):
        x, y = p.exterior.xy
        ax.fill(x, y, color=color, alpha=alpha, linewidth=0)

def hv_union_geom(front: pd.DataFrame, x_ref: float, y_ref: float,
                  x="voltage", y="separation"):
    rects = []
    for _, r in front.iterrows():
        xi = float(r[x]); yi = float(r[y])
        if xi <= x_ref or yi <= y_ref:
            continue
        rects.append(box(x_ref, y_ref, xi, yi))
    if not rects:
        return Polygon()
    return fix_geom(unary_union(rects))

def shiny_ball(ax, x, y, s=140, color=None, edgecolor="black", lw=1.0,
               highlight_alpha=0.35, zorder=20):
    if color is None:
        color = plt.rcParams["axes.prop_cycle"].by_key()["color"][0]

    ax.scatter([x], [y], s=s, color=color, edgecolors=edgecolor, linewidths=lw, zorder=zorder)

    fig = ax.figure
    offset = mtransforms.ScaledTranslation(-2/72., 2/72., fig.dpi_scale_trans)
    ax.scatter([x], [y],
               s=s * 0.22,
               color="white",
               alpha=highlight_alpha,
               linewidths=0,
               transform=ax.transData + offset,
               zorder=zorder + 10)

# -------------------------
# TOP row plotting
# ✅ batch0-batch7 全部按 colors 着色散点
# ✅ Pareto front 仍然只画 batch0 & batch7
# -------------------------
def plot_system(ax, json_path: Path, sys_name: str):
    df_json, max_batch, x_ref, y_ref = load_system_df(json_path)

    if sys_name not in BATCH0_BY_SYS:
        raise KeyError(f"Missing batch0 definition for {sys_name} in BATCH0_BY_SYS")

    # batch0 point
    x0, y0 = BATCH0_BY_SYS[sys_name]
    x0 = _normalize_potential(x0)
    y0 = float(y0)

    df0 = pd.DataFrame({
        "id": [-1],
        "voltage": [x0],
        "separation": [y0],
        "batch": [0],
    })

    # df_all: batch0 + json
    df_all = pd.concat([df0, df_json], ignore_index=True)

    # fronts/hv for b=0..max_batch
    fronts, hv_geoms = {}, {}
    for b in range(0, max_batch + 1):
        cum = df_all[df_all["batch"] <= b].copy()
        fronts[b] = pareto_front_max(cum)
        hv_geoms[b] = hv_union_geom(fronts[b], x_ref, y_ref)

    # colors: index == batch (0..max_batch)
    cmap = plt.cm.rainbow
    colors = [cmap(i / max(1, max_batch)) for i in range(max_batch + 1)]
    if REVERSE_RAINBOW:
        colors = colors[::-1]

    early_color = colors[0]          # batch0
    last_color  = colors[max_batch]  # batch7

    # ✅ scatter: batch0..batch7 全部用 colors[b]
    for b in range(0, max_batch + 1):
        c = colors[b]
        dfb = df_all[df_all["batch"] == b]
        ax.scatter(dfb["voltage"], dfb["separation"],
                   s=S_SCATTER * (1.2 if b == 0 else 1.0),
                   color=c, alpha=SCATTER_ALPHA,
                   edgecolors="black", linewidths=0.8, zorder=110 if b != 0 else 200)

    # HV fill/boundary: batch0 vs batch7
    hv0 = fix_geom(hv_geoms[0])
    hvl = fix_geom(hv_geoms[max_batch])

    fill_geom(ax, hv0, early_color, alpha=AREA_ALPHA)
    if max_batch > 0:
        delta_last = fix_geom(hvl.difference(hv0))
        fill_geom(ax, delta_last, last_color, alpha=AREA_ALPHA)

    draw_geom_boundary(ax, hv0, early_color, alpha=HV_LINE_ALPHA_EARLY, lw=HV_LINE_LW)
    draw_geom_boundary(ax, hvl, last_color,  alpha=HV_LINE_ALPHA_LAST,  lw=HV_LINE_LW)

    # Pareto front: only batch0 & batch7
    f0 = fronts[0]
    ax.scatter(f0["voltage"], f0["separation"],
               s=PARETO_POINT_SIZE, color=early_color,
               edgecolors="black", linewidths=0.8,
               alpha=1.0, zorder=220)

    f_last = fronts[max_batch]
    ax.scatter(f_last["voltage"], f_last["separation"],
               s=PARETO_POINT_SIZE, color=last_color,
               edgecolors="black", linewidths=0.8,
               alpha=PARETO_POINT_ALPHA, zorder=130)
    ax.plot(f_last["voltage"], f_last["separation"],
            linewidth=PARETO_LINE_LW, markersize=PARETO_MARKERSIZE,
            color=last_color, alpha=PARETO_LINE_ALPHA_LAST, linestyle="--")

    # ticks + frame
    from matplotlib.ticker import MultipleLocator
    ax.yaxis.set_major_locator(MultipleLocator(1))
    ax.xaxis.set_major_locator(MultipleLocator(0.05))
    ax.tick_params(axis="both", which="both", labelsize=24)
    for t in ax.get_xticklabels() + ax.get_yticklabels():
        t.set_fontname("Arial")
        t.set_fontweight("bold")
    for side in ["left", "right", "top", "bottom"]:
        ax.spines[side].set_linewidth(FRAME_LW)

# -------------------------
# BOTTOM row plotting (保持不变)
# -------------------------
def compute_hv_series(json_path: Path, sys_name: str):
    df_json, max_batch, x_ref, y_ref = load_system_df(json_path)

    if sys_name not in BATCH0_BY_SYS:
        raise KeyError(f"Missing batch0 definition for {sys_name} in BATCH0_BY_SYS")

    x0, y0 = BATCH0_BY_SYS[sys_name]
    x0 = _normalize_potential(x0)
    y0 = float(y0)

    df0 = pd.DataFrame({
        "id": [-1],
        "voltage": [x0],
        "separation": [y0],
        "batch": [0],
    })

    df_all = pd.concat([df0, df_json], ignore_index=True)
    batches = np.arange(0, max_batch + 1, dtype=int)

    hv = []
    for b in batches:
        cum = df_all[df_all["batch"] <= b].copy()
        front = pareto_front_max(cum)
        hv_geom = hv_union_geom(front, x_ref, y_ref)
        hv.append(float(fix_geom(hv_geom).area))

    hv = np.asarray(hv, dtype=float)

    if HV_MODE == "absolute":
        y = hv
    elif HV_MODE == "gain":
        y = hv - hv[0]
        y[0] = 0.0
    elif HV_MODE == "normalized":
        denom = (hv[-1] - hv[0])
        y = (hv - hv[0]) / denom if abs(denom) > 1e-12 else np.zeros_like(hv)
        y[0] = 0.0
    else:
        raise ValueError("HV_MODE must be one of: 'gain', 'absolute', 'normalized'")

    return batches, y, hv

def plot_hv_panel(ax, json_path: Path, sys_name: str):
    batches, y, _ = compute_hv_series(json_path, sys_name)

    main_color = "#FF42AD"
    ax.plot(batches, y, linewidth=LINE_LW, alpha=LINE_ALPHA, color=main_color, zorder=10)

    for i, b in enumerate(batches):
        shiny_ball(ax, b, y[i], s=SCATTER_S*1.2, color=main_color, lw=1.0, zorder=20)

    shiny_ball(ax, batches[0],  y[0],  s=HIGHLIGHT_S, color=main_color, lw=0.0, zorder=30)
    shiny_ball(ax, batches[-1], y[-1], s=HIGHLIGHT_S, color=main_color, lw=0.0, zorder=31)

    ax.set_xticks(batches)
    ax.tick_params(axis="both", which="both", labelsize=24)
    for t in ax.get_xticklabels() + ax.get_yticklabels():
        t.set_fontname("Arial")
        t.set_fontweight("bold")

    ax.margins(x=0.05, y=0.10)
    for side in ["left", "right", "top", "bottom"]:
        ax.spines[side].set_linewidth(FRAME_LW)

# -------------------------
# 2×4 figure
# -------------------------
ncol = len(systems)
fig_w = TOP_FIGSIZE_PER_PANEL[0] * ncol
fig_h = TOP_FIGSIZE_PER_PANEL[1] * 2

fig, axs = plt.subplots(
    2, ncol,
    figsize=(fig_w, fig_h),
    gridspec_kw={"height_ratios": [1.15, 1.00]}
)

for j, (p, sys_name) in enumerate(systems):
    ax_top = axs[0, j]
    ax_bot = axs[1, j]

    plot_system(ax_top, p, sys_name)
    plot_hv_panel(ax_bot, p, sys_name)

    ax_top.set_title(sys_name, fontname="Arial", fontweight="bold", fontsize=30, pad=14)
    ax_top.set_xlabel("")
    ax_top.set_ylabel("")

    ax_bot.set_title(sys_name, fontname="Arial", fontweight="bold", fontsize=30, pad=14)
    ax_bot.set_xlabel("")
    ax_bot.set_ylabel("")

ROW_HPAD = 10.0
fig.tight_layout(rect=[0.11, 0.06, 0.99, 0.98], h_pad=ROW_HPAD)

top_axes = list(axs[0, :].ravel())
bot_axes = list(axs[1, :].ravel())
bbox_top = mtransforms.Bbox.union([ax.get_position() for ax in top_axes])
bbox_bot = mtransforms.Bbox.union([ax.get_position() for ax in bot_axes])

pad = 0.02
y_center_top = 0.5 * (bbox_top.y0 + bbox_top.y1)
x_center_top = 0.5 * (bbox_top.x0 + bbox_top.x1)
x_left_top   = bbox_top.x0

fig.text(
    x_left_top - 1.5*pad, y_center_top,
    "Separation Coefficient (@ -50 mA/cm²)",
    rotation="vertical", va="center", ha="center",
    fontname="Arial", fontweight="bold", fontsize=30
)

y_between = 0.5 * (bbox_top.y0 + bbox_bot.y1)
fig.text(
    x_center_top, y_between + 1.2*pad,
    "Potential (@ -10 mA/cm², vs. RHE, V)",
    va="center", ha="center",
    fontname="Arial", fontweight="bold", fontsize=30
)

y_center_bot = 0.5 * (bbox_bot.y0 + bbox_bot.y1)
x_center_bot = 0.5 * (bbox_bot.x0 + bbox_bot.x1)
x_left_bot   = bbox_bot.x0

y_label = {
    "gain": "Improvement of Hypervolume",
    "absolute": "Hypervolume",
    "normalized": "Normalized Hypervolume Gain (0–1)"
}[HV_MODE]

fig.text(
    x_left_bot - 1.5*pad, y_center_bot,
    y_label,
    rotation="vertical", va="center", ha="center",
    fontname="Arial", fontweight="bold", fontsize=30
)

fig.text(
    x_center_bot, 2.5*pad,
    "Batch Number",
    va="center", ha="center",
    fontname="Arial", fontweight="bold", fontsize=30
)

plt.show()
fig.savefig("eight_panels_2x4.png", dpi=600, bbox_inches="tight", pad_inches=0.02)
```

```python
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D
from matplotlib.legend_handler import HandlerBase
import matplotlib as mpl
from matplotlib.colors import to_hex

# ============ 字体：Arial ============
mpl.rcParams["font.family"] = "Arial"

# ============ 与主图一致的设置 ============
MAX_BATCH = 7
REVERSE_RAINBOW = True
cmap = plt.cm.rainbow

# ✅ 与主图一致：batch0..batch7 颜色
colors = [cmap(i / max(1, MAX_BATCH)) for i in range(MAX_BATCH + 1)]  # index = 0..7
if REVERSE_RAINBOW:
    colors = colors[::-1]

batch_color = {b: colors[b] for b in range(0, MAX_BATCH + 1)}

# （可选）打印核对
print("Batch -> color")
for b in range(0, MAX_BATCH + 1):
    print(f"Batch {b}: {to_hex(batch_color[b])}  RGBA={batch_color[b]}")

# ============ 自定义：两条虚线“竖着叠放”的 legend handler ============
class ParetoHandle:
    pass

class HandlerTwoDashedLines(HandlerBase):
    def __init__(self, colors, lw=2.0, ls="--", **kwargs):
        super().__init__(**kwargs)
        self.colors = colors  # (top, bottom)
        self.lw = lw
        self.ls = ls

    def create_artists(self, legend, orig_handle, x0, y0, width, height, fontsize, trans):
        x1 = x0 + width
        y_top = y0 + 0.72 * height
        y_bot = y0 + 0.28 * height

        line_top = Line2D([x0, x1], [y_top, y_top],
                          linestyle=self.ls, linewidth=self.lw,
                          color=self.colors[0], transform=trans)
        line_bot = Line2D([x0, x1], [y_bot, y_bot],
                          linestyle=self.ls, linewidth=self.lw,
                          color=self.colors[1], transform=trans)
        return [line_top, line_bot]

# ============ 生成 legend ============
handles, labels = [], []

for b in range(0, MAX_BATCH + 1):
    h = Line2D([0], [0],
               marker="o", linestyle="None",
               markerfacecolor=batch_color[b],
               markeredgecolor="black",
               markeredgewidth=1.2,
               markersize=10)
    handles.append(h)
    labels.append(f"Batch {b}")

pareto = ParetoHandle()
handles.append(pareto)
labels.append("Pareto Front")

# ============ 画布只放 legend ============
fig = plt.figure(figsize=(24, 1.2), dpi=300)
ax = fig.add_axes([0, 0, 1, 1])
ax.axis("off")

leg = ax.legend(
    handles, labels,
    loc="center",
    ncol=len(labels),
    frameon=True,
    fancybox=False,
    edgecolor="black",
    framealpha=1.0,
    borderpad=0.4,
    handletextpad=0.6,
    columnspacing=1.1,
    prop={"family": "Arial", "weight": "bold", "size": 14},
    handler_map={
        ParetoHandle: HandlerTwoDashedLines(
            colors=(batch_color[0], batch_color[7]),  # 上=Batch0，下=Batch7
            lw=2.0,
            ls="--"
        )
    }
)

# ✅ Pareto Front 文字加粗
leg.get_texts()[-1].set_fontweight("bold")

leg.get_frame().set_linewidth(1.5)

out_png = "custom_legend_batch0_7_rainbow_arial.png"
out_svg = "custom_legend_batch0_7_rainbow_arial.svg"
fig.savefig(out_png, bbox_inches="tight", pad_inches=0.05)
fig.savefig(out_svg, bbox_inches="tight", pad_inches=0.05)
plt.close(fig)

print("Saved:", out_png, out_svg)



```
