```python
import pandas as pd 
import json 
import numpy as np

BATCH_NUM = 10
```

```python
exp_data = []
for BATCH_NUM_tmp in range(BATCH_NUM+1):
    with open(f"./all_experiment_{BATCH_NUM_tmp}.json", "r") as f:
        exp_data.append(pd.DataFrame(json.load(f)))
        
exp_data = pd.concat(exp_data).sort_values(by=["substance1", "substance2", "substance3", "substance4", "substance5","volume1","volume2","volume3","volume4"]).reset_index(drop=True)

if not (exp_data.index.astype("int") == np.arange(len(exp_data))).all():
    raise ValueError("the index of the expeirments is not right")

# add noise from Pt experiments
exp_data["variance_S"] = exp_data["variance_S"] + 1.6
exp_data["variance_current"] = exp_data["variance_current"] + 198.53
exp_data["score"] =  exp_data["mean_S"] + exp_data["mean_current"]/40 # 40 mA = 1 separation coefficient
```

```python
exp_data = exp_data.sort_values(by = ["score"], ascending = False, ignore_index=True) 
exp_data
```

```python
# Drop duplicates based on 'substance1' and 'substance2' to ensure unique pairs
unique_system = exp_data.drop_duplicates(subset=['substance1', 'substance2'])

# Select the top 3 unique pairs
selected_systems = unique_system.head(3)

selected_systems
```

```python
selected_systems.to_csv("./selected_systems.csv", index = False)
```
