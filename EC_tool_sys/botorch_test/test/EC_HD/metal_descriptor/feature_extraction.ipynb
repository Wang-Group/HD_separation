```python
import numpy as np
import pandas as pd
from matplotlib import pyplot as plt
data = pd.read_excel("./descriptor.xlsx",header=None)
```

```python
current = []
current_std = []
k_HD = []
k_HD_std = []
for i in data.values[:,1]:
    c_temp = str(i).split("±")
    if len(c_temp) >1:
        current.append(float(c_temp[0]))
        current_std.append(float(c_temp[1]))
    else:
        current.append(float(c_temp[0]))
        current_std.append(None)

for i in data.values[:,2]:
    c_temp = str(i).split("±")
    if len(c_temp) >1:
        k_HD.append(float(c_temp[0]))
        k_HD_std.append(float(c_temp[1]))
    else:
        k_HD.append(float(c_temp[0]))
        k_HD_std.append(None)
```

```python
d = {}
d["metal"] = data.values[:,0]
d["current"] = current
d["current_std"] = current_std
d["k_HD"] = k_HD
d["k_HD_std"] = k_HD_std
d = pd.DataFrame(d)
```

```python
d.to_csv("metal_descriptor.csv")
```
