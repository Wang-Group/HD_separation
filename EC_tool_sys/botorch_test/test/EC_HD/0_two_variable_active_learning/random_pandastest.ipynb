```python
import pandas as pd
import numpy as np
import random

# Sample DataFrame
data = {'A': np.arange(0, 10),
        'B': np.random.randn(10)}

df = pd.DataFrame(data)
print(df)
```

```python
df.iloc[5]
```

```python
# Shuffle the index
index_list = df.index.tolist()
random.shuffle(index_list)
print(100*"=")
# Reassign the shuffled index to the DataFrame
df = df.reindex(index=index_list)

print(df)
```
